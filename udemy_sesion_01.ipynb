```python
word:str = 'curso de python'
print (word.capitalize())
print (word.title())

#Eliminamos espacios
words = '           hola             andres      '
print(words.strip())
print(words.lstrip())
print(words.rstrip())

#Remplazar un texto de un String
text:str = 'Hola xgeospace'
new_text= text.replace('xgeospace','Felix')
print(new_text)

#Convertimos un String a una lista
text:str = 'andres, jaime, pepito, ruben, euclides'
datalist= text.split(',')
print(datalist)
print(datalist[3])
ruben = datalist[3]
print(ruben)
print('==========================================================')

#Convertimos una lista a un String
data:list = ['andres', ' jaime', ' pepito', ' ruben', ' euclides']
print(data)
text = ' - '.join(data)
print(text)


```
